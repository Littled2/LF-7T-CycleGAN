## Patch Extractor

This file is used to generate patch overlays on images to be used in visualisations.



#### How to use:

1. Copy the `.png` images that to be patched into a directory called `source-images/`. This folder must be in the same directory as this notebook file.
2. Adjust the position of the zoomed area by increasing or decreasing the values of `PATCH_X` and `PATCH_Y`.
3. Run the below cells. The generated images will appear in the same directory as this notebook with the suffix: `_inset.png`.

```python
import matplotlib.pyplot as plt
from pathlib import Path
import numpy as np
from PIL import Image
```

```python
# Define the position of the zoomed area
PATCH_X = 55
PATCH_Y = 25

IMAGE_DIR   = "source-images"
PATCH_X_DIM = 80
PATCH_Y_DIM = 80

# For each image in the source-images directory, create an inset zoomed area.
for path in Path(IMAGE_DIR).glob("*.png"):
    img = np.array(Image.open(path))

    fig, ax = plt.subplots()
    ax.imshow(img)
    ax.axis("off")
    
    # Create a rectngle around the zoomed patch
    ax.add_patch(plt.Rectangle((PATCH_X, PATCH_Y), PATCH_X_DIM, PATCH_Y_DIM, linewidth=2, edgecolor="white", facecolor="none"))

    # Create an inset axes and display the zoomed area
    ax_insert = ax.inset_axes([0.6, 0.0, 0.4, 0.4])
    ax_insert.imshow(img[PATCH_Y:PATCH_Y + PATCH_Y_DIM, PATCH_X:PATCH_X + PATCH_X_DIM])
    ax_insert.set_xticks([]); ax_insert.set_yticks([])

    # Draw outline around the inset area
    for spine in ax_insert.spines.values():
        spine.set_edgecolor("white"); spine.set_linewidth(2)

    # Save image with inset
    plt.savefig(path.stem + "_inset.png", dpi=150, bbox_inches="tight", pad_inches=0)
    plt.close()
```
