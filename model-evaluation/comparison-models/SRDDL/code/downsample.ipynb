# Downsampler

This notebook allows for 256x256 images to be downsampled to 128x128 for use by SRDDL.

```python
# Imports

import numpy as np
from pathlib import Path
from skimage.transform import resize
from tqdm import tqdm
```

```python
# Paths to input 256x256 scans and output directory must be specified

INPUT_256_DIR = Path("G:\\LF-7T-CycleGAN\\pytorch-cyclegan-and-pix2pix\\results\\SRDDL-evaluation-dataset\\T1w\\LR-256")
OUTPUT_128_DIR = Path("G:\\LF-7T-CycleGAN\\pytorch-cyclegan-and-pix2pix\\results\\SRDDL-evaluation-dataset\\T1w\\LR")
```

```python
# Ensure output dir exists
OUTPUT_128_DIR.mkdir(parents=True, exist_ok=True)

# For each file, downsample and save as .npy
files = list(INPUT_256_DIR.glob("*.npy"))

# Resize images and show progress
for file in tqdm(files, desc="Downsampling sand saving"):
    file_data = np.load(file)
    downsampled = resize(file_data, (128, 128), anti_aliasing=True, preserve_range=True).astype(file_data.dtype)
    np.save(OUTPUT_128_DIR / file.name, downsampled)
```
